## PROFILING THE DATA
### importing all libraries 

```python
import pandas as pd
import numpy as np
import json

import matplotlib.pyplot as plt
import seaborn as sns

from pathlib import Path

pd.set_option("display.max_columns", None)
pd.set_option("display.max_rows", 100)
pd.set_option("display.max_colwidth", 100)

sns.set_theme(style="whitegrid")
```

## Locate your raw data

```python
BASE_DIR = Path.cwd().parent

RAW_DIR = BASE_DIR / "data" / "raw"

DETAILS_FILE = RAW_DIR / "movie_details.json"
CREDITS_FILE = RAW_DIR / "movie_credits.json"

print("Project directory:", BASE_DIR)
print("Raw data directory:", RAW_DIR)

print("Movie details exists:", DETAILS_FILE.exists())
print("Movie credits exists:", CREDITS_FILE.exists())
```

## Load the JSON files


```python
with open(DETAILS_FILE, "r", encoding="utf-8") as file:
    movie_details = json.load(file)

with open(CREDITS_FILE, "r", encoding="utf-8") as file:
    movie_credits = json.load(file)

print("Movie details records:", len(movie_details))
print("Movie credits records:", len(movie_credits))
```

## Convert JSON into DataFrames

```python
movies_df = pd.json_normalize(movie_details)

credits_df = pd.json_normalize(movie_credits)

print("Movie details shape:", movies_df.shape)
print("Movie credits shape:", credits_df.shape)
```

```python
movies_df.columns = movies_df.columns.str.replace(".", "_", regex=False)  
movies_df.head()  
```

```python
credits_df.columns = credits_df.columns.str.replace(".", "_", regex=False)  
credits_df.head()
```

```python
column_types = {
    "integer": [
        "id", "budget", "revenue", "runtime", "vote_count",
        "belongs_to_collection_id"
    ],

    "float": [
        "popularity", "vote_average"
    ],

    "boolean": [
        "adult", "video"
    ],

    "datetime": [
        "release_date"
    ],

    "text": [
        "homepage", "imdb_id", "original_language",
        "original_title", "overview", "status",
        "tagline", "title", "backdrop_path", "poster_path",
        "belongs_to_collection_name",
        "belongs_to_collection_poster_path",
        "belongs_to_collection_backdrop_path"
    ],

    "nested": [
        "genres", "origin_country", "production_companies",
        "production_countries", "spoken_languages",
        "belongs_to_collection"
    ],

    "investigate": [
        "softcore"
    ]
}
```

```python
classification_records = []

for column in movies_df.columns:

    expected_type = "unclassified"

    for category, columns in column_types.items():

        if column in columns:
            expected_type = category
            break

    classification_records.append({
        "column": column,
        "actual_dtype": str(movies_df[column].dtype),
        "expected_type": expected_type,
        "missing_count": movies_df[column].isna().sum(),
      
    })

classification_df = pd.DataFrame(classification_records)

classification_df
```

```python
quarantine_dir = BASE_DIR / "data" / "quarantine"

quarantine_dir.mkdir(parents=True, exist_ok=True)

softcore_quarantine = movies_df[
    ["id", "title", "softcore"]
].copy()

softcore_quarantine["quarantine_reason"] = (
    "Unverified field meaning and expected data type"
)

softcore_quarantine.to_csv(
    quarantine_dir / "softcore_review.csv",
    index=False
)

print("Softcore records quarantined:",
      len(softcore_quarantine))
```

```python
missing_values = pd.DataFrame({
    "Missing_Count": movies_df.isnull().sum(),
    "Missing_Percentage": (
        movies_df.isnull().sum() / len(movies_df) * 100
    )
})

missing_values = missing_values.sort_values(
    by="Missing_Percentage",
    ascending=False
)

missing_values
```

```python
invalid_records = []

numeric_columns = (
    column_types["integer"] + column_types["float"]
)

for column in numeric_columns:

    if column not in movies_df.columns:
        continue

    converted = pd.to_numeric(
        movies_df[column],
        errors="coerce"
    )

    invalid_mask = (
        movies_df[column].notna() &
        converted.isna()
    )

    for index in movies_df.index[invalid_mask]:

        invalid_records.append({
            "row_index": index,
            "movie_id": movies_df.loc[index, "id"],
            "column": column,
            "original_value": str(
                movies_df.loc[index, column]
            ),
            "expected_type": (
                "integer" if column in column_types["integer"]
                else "float"
            ),
            "quarantine_reason": "Invalid numeric value"
        })

invalid_values_df = pd.DataFrame(invalid_records)

invalid_values_df.to_csv(
    quarantine_dir / "invalid_values.csv",
    index=False
)

print("Invalid values found:", len(invalid_values_df))
```
